```python
# -------------------------------------------------------
# Computer Vision Research Pipeline
# CIFAR-10 + ResNet-18
# Metode de augmentare: Baseline, MixUp, CutMix, RandAugment
# -------------------------------------------------------

# Importarea bibliotecilor necesare
import torch
import torchvision
import torchvision.transforms as transforms
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from sklearn.metrics import confusion_matrix


# Selectam automat GPU daca este disponibil, altfel CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"\n{'='*50}")
print(f"  Dispozitiv utilizat: {device}")
print(f"{'='*50}\n")


# Transformari pentru antrenare: Random Crop, Horizontal Flip, Normalizare
transform_train = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465),
                         (0.2023, 0.1994, 0.2010))
])

# Transformari pentru testare: doar Normalizare
transform_test = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465),
                         (0.2023, 0.1994, 0.2010))
])


# Descarcam si incarcam CIFAR-10 (50.000 imagini antrenare, 10.000 testare)
trainset = torchvision.datasets.CIFAR10(
    root='./data', train=True, download=True, transform=transform_train)

testset = torchvision.datasets.CIFAR10(
    root='./data', train=False, download=True, transform=transform_test)

trainloader = torch.utils.data.DataLoader(
    trainset, batch_size=128, shuffle=True, num_workers=2)

testloader = torch.utils.data.DataLoader(
    testset, batch_size=128, shuffle=False, num_workers=2)

classes = trainset.classes
print(f"Clase CIFAR-10: {classes}\n")


# Functie pentru afisarea imaginilor denormalizate
def imshow(img):
    img = img / 2 + 0.5
    npimg = img.numpy()
    plt.figure(figsize=(10, 5))
    plt.imshow(np.transpose(npimg, (1, 2, 0)))
    plt.axis('off')
    plt.show()

# Afisam 8 imagini exemplu din setul de antrenare
dataiter = iter(trainloader)
images, labels = next(dataiter)
imshow(torchvision.utils.make_grid(images[:8].cpu()))
print("Etichete:", ' '.join(f'{classes[labels[j]]}' for j in range(8)))


# Functia generica de antrenare: ruleaza epocile, calculeaza loss si acuratete
def train_model(model, trainloader, criterion, optimizer, epochs=20):
    train_losses = []
    train_accuracies = []

    for epoch in range(epochs):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        loop = tqdm(trainloader)

        for inputs, labels in loop:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            running_loss += loss.item()
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
            loop.set_description(f"Epoch [{epoch+1}/{epochs}]")

        epoch_loss = running_loss / len(trainloader)
        epoch_acc = 100. * correct / total
        train_losses.append(epoch_loss)
        train_accuracies.append(epoch_acc)
        print(f"  Epoca {epoch+1:>2}/{epochs} | Loss: {epoch_loss:.4f} | Acuratete: {epoch_acc:.2f}%")

    return train_losses, train_accuracies


# Evalueaza modelul pe setul de test; returneaza acuratete, predictii, etichete
def evaluate_model(model, testloader):
    model.eval()
    correct = 0
    total = 0
    all_preds = []
    all_labels = []

    with torch.no_grad():
        for inputs, labels in testloader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            _, predicted = outputs.max(1)
            total += labels.size(0)
            correct += predicted.eq(labels).sum().item()
            all_preds.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

    accuracy = 100. * correct / total
    return accuracy, all_preds, all_labels


# --- BASELINE: ResNet-18 (He et al., 2015) ---
# Retea reziduala cu 18 straturi, fara augmentari speciale
print(f"\n{'='*50}")
print("  [1/4] ANTRENARE BASELINE - ResNet-18")
print(f"{'='*50}")

model = torchvision.models.resnet18(weights=None)
model.fc = nn.Linear(model.fc.in_features, 10)
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

baseline_losses, baseline_accs = train_model(
    model, trainloader, criterion, optimizer, epochs=20)

baseline_test_acc, baseline_preds, baseline_labels = evaluate_model(model, testloader)
print(f"\n  >>> Baseline Test Accuracy: {baseline_test_acc:.2f}%")


# --- MIXUP: (Zhang et al., 2018) ---
# Combina doua imagini si etichetele lor folosind distributia Beta
# Formula: x_mixed = lam * x_i + (1 - lam) * x_j
def mixup_data(x, y, alpha=1.0):
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1
    batch_size = x.size()[0]
    index = torch.randperm(batch_size).to(device)
    mixed_x = lam * x + (1 - lam) * x[index, :]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

print(f"\n{'='*50}")
print("  [2/4] ANTRENARE MIXUP")
print(f"{'='*50}")

model_mixup = torchvision.models.resnet18(weights=None)
model_mixup.fc = nn.Linear(model_mixup.fc.in_features, 10)
model_mixup = model_mixup.to(device)

criterion_mixup = nn.CrossEntropyLoss()
optimizer_mixup = optim.Adam(model_mixup.parameters(), lr=0.001)

mixup_train_losses = []
mixup_train_accuracies = []
epochs = 20

for epoch in range(epochs):
    model_mixup.train()
    running_loss = 0.0
    correct = 0
    total = 0
    loop = tqdm(trainloader)

    for inputs, labels in loop:
        inputs, labels = inputs.to(device), labels.to(device)
        inputs, targets_a, targets_b, lam = mixup_data(inputs, labels, alpha=1.0)
        optimizer_mixup.zero_grad()
        outputs = model_mixup(inputs)
        loss = mixup_criterion(criterion_mixup, outputs, targets_a, targets_b, lam)
        loss.backward()
        optimizer_mixup.step()

        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += (lam * predicted.eq(targets_a).sum().item()
                     + (1 - lam) * predicted.eq(targets_b).sum().item())
        loop.set_description(f"MixUp Epoch [{epoch+1}/{epochs}]")

    epoch_loss = running_loss / len(trainloader)
    epoch_acc = 100. * correct / total
    mixup_train_losses.append(epoch_loss)
    mixup_train_accuracies.append(epoch_acc)
    print(f"  Epoca {epoch+1:>2}/{epochs} | Loss: {epoch_loss:.4f} | Acuratete: {epoch_acc:.2f}%")

mixup_test_acc, _, _ = evaluate_model(model_mixup, testloader)
print(f"\n  >>> MixUp Test Accuracy: {mixup_test_acc:.2f}%")


# --- CUTMIX: (Yun et al., 2019) ---
# Taie o portiune dreptunghiulara dintr-o imagine si o inlocuieste cu alta
# Etichetele se ajusteaza proportional cu aria regiunii decupate
def rand_bbox(size, lam):
    W = size[2]
    H = size[3]
    cut_rat = np.sqrt(1. - lam)
    cut_w = int(W * cut_rat)
    cut_h = int(H * cut_rat)
    cx = np.random.randint(W)
    cy = np.random.randint(H)
    bbx1 = np.clip(cx - cut_w // 2, 0, W)
    bby1 = np.clip(cy - cut_h // 2, 0, H)
    bbx2 = np.clip(cx + cut_w // 2, 0, W)
    bby2 = np.clip(cy + cut_h // 2, 0, H)
    return bbx1, bby1, bbx2, bby2

print(f"\n{'='*50}")
print("  [3/4] ANTRENARE CUTMIX")
print(f"{'='*50}")

model_cutmix = torchvision.models.resnet18(weights=None)
model_cutmix.fc = nn.Linear(model_cutmix.fc.in_features, 10)
model_cutmix = model_cutmix.to(device)

criterion_cutmix = nn.CrossEntropyLoss()
optimizer_cutmix = optim.Adam(model_cutmix.parameters(), lr=0.001)

cutmix_train_losses = []
cutmix_train_accuracies = []
epochs = 20

for epoch in range(epochs):
    model_cutmix.train()
    running_loss = 0.0
    correct = 0
    total = 0
    loop = tqdm(trainloader)

    for inputs, labels in loop:
        inputs, labels = inputs.to(device), labels.to(device)
        lam = np.random.beta(1.0, 1.0)
        rand_index = torch.randperm(inputs.size()[0]).to(device)
        target_a = labels
        target_b = labels[rand_index]
        bbx1, bby1, bbx2, bby2 = rand_bbox(inputs.size(), lam)
        inputs[:, :, bbx1:bbx2, bby1:bby2] = inputs[rand_index, :, bbx1:bbx2, bby1:bby2]
        lam = 1 - ((bbx2 - bbx1) * (bby2 - bby1) / (inputs.size()[-1] * inputs.size()[-2]))
        optimizer_cutmix.zero_grad()
        outputs = model_cutmix(inputs)
        loss = lam * criterion_cutmix(outputs, target_a) + (1 - lam) * criterion_cutmix(outputs, target_b)
        loss.backward()
        optimizer_cutmix.step()

        running_loss += loss.item()
        _, predicted = outputs.max(1)
        total += labels.size(0)
        correct += (lam * predicted.eq(target_a).sum().item()
                     + (1 - lam) * predicted.eq(target_b).sum().item())
        loop.set_description(f"CutMix Epoch [{epoch+1}/{epochs}]")

    epoch_loss = running_loss / len(trainloader)
    epoch_acc = 100. * correct / total
    cutmix_train_losses.append(epoch_loss)
    cutmix_train_accuracies.append(epoch_acc)
    print(f"  Epoca {epoch+1:>2}/{epochs} | Loss: {epoch_loss:.4f} | Acuratete: {epoch_acc:.2f}%")

cutmix_test_acc, _, _ = evaluate_model(model_cutmix, testloader)
print(f"\n  >>> CutMix Test Accuracy: {cutmix_test_acc:.2f}%")


# --- RANDAUGMENT: (Cubuk et al., 2020) ---
# Aplica N transformari aleatoare cu magnitudine M din lista de augmentari
# Parametri: num_ops=2 (numarul de transformari), magnitude=9 (intensitate)
print(f"\n{'='*50}")
print("  [4/4] ANTRENARE RANDAUGMENT")
print(f"{'='*50}")

transform_train_randaug = transforms.Compose([
    transforms.RandomCrop(32, padding=4),
    transforms.RandomHorizontalFlip(),
    transforms.RandAugment(num_ops=2, magnitude=9),
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465),
                         (0.2023, 0.1994, 0.2010))
])

trainset_randaug = torchvision.datasets.CIFAR10(
    root='./data', train=True, download=True, transform=transform_train_randaug)

trainloader_randaug = torch.utils.data.DataLoader(
    trainset_randaug, batch_size=128, shuffle=True, num_workers=2)

model_randaug = torchvision.models.resnet18(weights=None)
model_randaug.fc = nn.Linear(model_randaug.fc.in_features, 10)
model_randaug = model_randaug.to(device)

criterion_randaug = nn.CrossEntropyLoss()
optimizer_randaug = optim.Adam(model_randaug.parameters(), lr=0.001)

randaug_losses, randaug_accs = train_model(
    model_randaug, trainloader_randaug, criterion_randaug, optimizer_randaug, epochs=20)

randaug_test_acc, _, _ = evaluate_model(model_randaug, testloader)
print(f"\n  >>> RandAugment Test Accuracy: {randaug_test_acc:.2f}%")


# Afisam rezultatele finale pentru toate metodele
methods = ["Baseline", "MixUp", "CutMix", "RandAugment"]
train_acc = [baseline_accs[-1], mixup_train_accuracies[-1],
             cutmix_train_accuracies[-1], randaug_accs[-1]]
test_acc = [baseline_test_acc, mixup_test_acc, cutmix_test_acc, randaug_test_acc]

print(f"\n{'='*55}")
print("          REZULTATE FINALE - COMPARATIE")
print(f"{'='*55}")
print(f"  {'Metoda':<14} | {'Train':>8} | {'Test':>8}")
print(f"  {'-'*14}-+-{'-'*8}-+-{'-'*8}")
for m, tr, te in zip(methods, train_acc, test_acc):
    print(f"  {m:<14} | {tr:>7.2f}% | {te:>7.2f}%")
print(f"{'='*55}\n")


# Grafic: Comparatie acuratete pe setul de test
x = range(len(methods))
plt.figure(figsize=(8, 5))
plt.bar(x, test_acc, color=['#1565c0', '#2e7d32', '#e65100', '#6a1b9a'])
plt.xticks(x, methods)
plt.ylabel("Test Accuracy (%)")
plt.title("Comparatie Metode de Augmentare - Acuratete Test")
plt.tight_layout()
plt.show()


# Grafic: Acuratete antrenare vs testare
width = 0.35
plt.figure(figsize=(8, 5))
plt.bar([i - width/2 for i in x], train_acc, width=width, label='Train Accuracy', color='#1565c0')
plt.bar([i + width/2 for i in x], test_acc, width=width, label='Test Accuracy', color='#4CAF50')
plt.xticks(x, methods)
plt.ylabel("Accuracy (%)")
plt.title("Acuratete Antrenare vs Test")
plt.legend()
plt.tight_layout()
plt.show()


# Matricea de confuzie pentru modelul Baseline
cm = confusion_matrix(baseline_labels, baseline_preds)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=classes, yticklabels=classes, cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Matrice de Confuzie - Baseline ResNet-18")
plt.tight_layout()
plt.show()
```
